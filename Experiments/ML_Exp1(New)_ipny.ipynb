Task 9: Fill missing Age values with the median.

```python
age_median = train_df['Age'].median()
train_df['Age'].fillna(age_median, inplace=True)
print(f"Missing 'Age' values after imputation: {train_df['Age'].isnull().sum()}")
```

Task 10: Filter surviving passengers (Survived == 1).

```python
survived_passengers_df = train_df[train_df['Survived'] == 1]
display(survived_passengers_df.head())
```

Task 11: Convert Fare column to a NumPy array.

```python
fare_array = train_df['Fare'].to_numpy()
print(f"Type of 'Fare' column after conversion: {type(fare_array)}")
display(fare_array[:5]) # Display the first 5 elements of the array
```

Task 13: Plot histogram of Age.

```python
plt.figure(figsize=(10, 6))
sns.histplot(train_df['Age'], bins=20, kde=True)
plt.title('Distribution of Age of Passengers')
plt.xlabel('Age')
plt.ylabel('Count')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

Task 14: Plot bar chart of Survival Count.

```python
plt.figure(figsize=(8, 5))
sns.countplot(x='Survived', hue='Survived', data=train_df, palette='viridis', legend=False)
plt.title('Survival Count (0 = Not Survived, 1 = Survived)')
plt.xlabel('Survived')
plt.ylabel('Count')
plt.xticks(ticks=[0, 1], labels=['Not Survived', 'Survived'])
plt.grid(axis='y', alpha=0.75)
plt.show()
```

Task 15: Plot scatter graph of Age vs Fare.

### Task 16: Handle missing values in 'Cabin' and 'Embarked' columns

```python
# Drop the 'Cabin' column due to a large number of missing values
train_df = train_df.drop(columns=['Cabin'])
print("Dropped 'Cabin' column.")

# Fill missing 'Embarked' values with the mode
embarked_mode = train_df['Embarked'].mode()[0]
train_df['Embarked'].fillna(embarked_mode, inplace=True)
print(f"Filled missing 'Embarked' values with: {embarked_mode}")

# Verify that all missing values have been handled
print("\nMissing values after cleaning:")
print(train_df.isnull().sum())
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Age', y='Fare', data=train_df, hue='Survived', palette='coolwarm', alpha=0.7)
plt.title('Scatter Plot of Age vs Fare (by Survival)')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```
