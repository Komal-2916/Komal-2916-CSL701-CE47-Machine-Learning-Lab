```python
# Now that the correct DataFrame is loaded, fill missing Age values with the median.
# Calculate the median Age
median_age = df['Age'].median()

# Fill missing Age values
df['Age'].fillna(median_age, inplace=True)

# Verify that there are no more missing Age values
display(df['Age'].isnull().sum())
```

Task 10: Filter surviving passengers (Survived == 1).

```python
survived_passengers = df[df['Survived'] == 1]
display(survived_passengers.head())
```

Task 11: Convert Fare column to a NumPy array.\

```python
fare_array = df['Fare'].to_numpy()
display(fare_array[:5]) # Displaying the first 5 elements to verify
```

Task 12: Calculate maximum and average Fare using np.max() and np.mean().

```python
max_fare = np.max(fare_array)
avg_fare = np.mean(fare_array)

print(f"Maximum Fare: {max_fare:.2f}")
print(f"Average Fare: {avg_fare:.2f}")
```

Task 13: Plot histogram of Age.

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['Age'], bins=20, kde=True)
plt.title('Distribution of Age')
plt.xlabel('Age')
plt.ylabel('Count')
plt.grid(True)
plt.show()
```

Task 14: Plot bar chart of Survival Count.

```python
plt.figure(figsize=(8, 5))
sns.countplot(x='Survived', data=df, palette='viridis')
plt.title('Survival Count (0 = Not Survived, 1 = Survived)')
plt.xlabel('Survival Status')
plt.ylabel('Number of Passengers')
plt.xticks(ticks=[0, 1], labels=['Not Survived', 'Survived'])
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

Task 15: Plot scatter graph of Age vs Fare.

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Age', y='Fare', data=df, hue='Survived', palette='viridis', alpha=0.7)
plt.title('Scatter Plot of Age vs Fare')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```
